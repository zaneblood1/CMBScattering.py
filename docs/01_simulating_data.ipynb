# Generating multi-frequency CMB + dust data

This guide simulates the data model of `cmb_scattering.simulate_dust` and plots the resulting maps.

Each of the three frequency channels observes the same CMB plus a rescaled copy of one dust field, plus its own noise. Everything is polarization only (E and B):

$$d_\nu = M_\nu B_\nu \,(f + a_\nu\, g) + n_\nu$$

| Symbol | Meaning | Statistics |
|---|---|---|
| $f$ | CMB, the same in every channel (maps are in $\mu K_{\rm CMB}$) | $f \sim N(0, C_f(r))$, with $C_f(r) = C_{\rm lensed\ scalar} + r\, C_{\rm tensor}$ |
| $g$ | dust at the 353 GHz reference frequency | $g \sim N(0, C_g)$, a separate $D_\ell$ power law for EE and BB |
| $a_\nu$ | dust frequency scaling, a modified blackbody relative to 353 GHz | set by $\beta_d$ and $T_d$ |
| $n_\nu$ | white noise, independent between channels | $n_\nu \sim N(0, C_n)$ |
| $M_\nu, B_\nu$ | mask and beam | the identity for now |


The Gaussian dust of this table is the default. The last section of this guide replaces it with a non-Gaussian dust map of the real sky from PySM.

## Why a large patch with coarse pixels

The tensor-to-scalar ratio $r$ sets the amplitude of primordial gravitational waves, which are the only primordial source of B-mode polarization. A gravitational wave decays once its wavelength is inside the horizon, so the B-modes it leaves behind sit on scales larger than the horizon at the time the polarization was made:

- **Recombination bump, $\ell \approx 80$.** The horizon at last scattering subtends about $2^\circ$.
- **Reionization bump, $\ell \lesssim 10$.** Photons rescatter after reionization, when the horizon is far larger.

Above $\ell \sim 150$ the tensor signal falls away, while its competitors grow: lensing B-modes behave like white noise of about $5\ \mu K$-arcmin at low $\ell$ and peak near $\ell \sim 1000$, and instrument noise is white too. Almost all the information on $r$ is therefore at $\ell \lesssim 150$.

A flat periodic patch of `nside` pixels of width $\theta_{\rm pix}$ only contains the multipoles

$$\ell_{\min} = \frac{2\pi}{{\tt nside}\ \theta_{\rm pix}}, \qquad \ell_{\rm Nyquist} = \frac{\pi}{\theta_{\rm pix}} .$$

At a fixed `nside` the number of Fourier modes is fixed, and the pixel size decides where they land. Small pixels spend them at high $\ell$, where there is no tensor signal. Coarse pixels move the same modes down to the scales that constrain $r$.

Two costs come with the larger patch. Dust is red ($D_\ell \propto \ell^{\alpha_d}$ with $\alpha_d < 0$), so it is brightest exactly where the tensor signal lives. And the flat-sky approximation degrades as the patch grows, so a patch of tens of degrees is a convenient toy geometry and not a faithful stand-in for a curved sky.


```python
import numpy as np
import jax.numpy as jnp
import matplotlib.pyplot as plt
from cmb_lensing.util import gen_ell_grid
from cmb_scattering.simulate_dust import (simulate_data, PLANCK_DUST_KNEE_EE,
                                          PLANCK_DUST_BB_TO_EE_AMPLITUDE_RATIO,
                                          DUST_REFERENCE_FREQUENCY_GHZ)

ARCMIN_PER_DEGREE = 60
DEGREES_PER_CIRCLE = 360

def patch_multipoles(nside, theta_pix):
    #(patch width in degrees, lowest multipole, Nyquist multipole) of a flat patch
    width_deg = nside * theta_pix / ARCMIN_PER_DEGREE
    ell_min = DEGREES_PER_CIRCLE / width_deg
    ell_nyquist = DEGREES_PER_CIRCLE * ARCMIN_PER_DEGREE / (2 * theta_pix)
    return width_deg, ell_min, ell_nyquist

nside = 128
#a fine-pixel patch, typical for lensing work, against the coarse patch used below
theta_pix_fine = 2.5
theta_pix = 20.0

for label, pixel in (("fine", theta_pix_fine), ("coarse", theta_pix)):
    width_deg, ell_min, ell_nyquist = patch_multipoles(nside, pixel)
    print(f"{label:>6}: {pixel:4.1f}' pixels, {width_deg:4.1f} deg patch, "
          f"ell from {ell_min:.0f} to {ell_nyquist:.0f}")
```

With 2.5' pixels the lowest multipole is already 68, so the recombination bump is covered by a handful of modes and the reionization bump not at all. With 20' pixels the same grid reaches down to $\ell \approx 8$.

## Simulating a data set

`simulate_data` takes the patch geometry, a seed and the model parameters. The values below follow the BICEP/Keck 2018 analysis ([BK18](https://arxiv.org/abs/2110.00483)) for the B-mode dust and [Planck 2018 XI](https://arxiv.org/abs/1801.04945) for the E-mode dust:

- `r = 0.03`, just under the BK18 limit $r < 0.036$ (95%).
- BB dust: $A_d = 4.4\ \mu K^2$ at 353 GHz and $\ell = 80$, $\alpha_d = -0.66$ (BK18 best fit).
- EE dust: BK18 does not fit it, so the amplitude is $A_d / 0.53$ (Planck's BB/EE ratio) and the slope is Planck's $-0.42$.
- $\beta_d = 1.49$ (BK18), with $T_d = 19.6$ K as the default.
- In this toy example we use $3\ \mu K$-arcmin of polarization noise in every channel. In reality, BK18's map depths are 2.8, 2.8 and 8.8 $\mu K$-arcmin at 95, 150 and 220 GHz.


```python
r = 0.03
dust_amplitude_bb = 4.4
dust_knee_bb = -0.66
dust_amplitude_ee = dust_amplitude_bb / PLANCK_DUST_BB_TO_EE_AMPLITUDE_RATIO
dust_knee_ee = PLANCK_DUST_KNEE_EE
frequency_exponent = 1.49
uk_arcmin = 3.0
master_seed = 67

sim = simulate_data(nside, theta_pix, master_seed, r, frequency_exponent,
                    dust_amplitude_ee = dust_amplitude_ee, dust_knee_ee = dust_knee_ee,
                    dust_amplitude_bb = dust_amplitude_bb, dust_knee_bb = dust_knee_bb,
                    uk_arcmin = uk_arcmin,
                    frequencies = (95.0, 150.0, 220.0, 353.0))

frequencies = np.asarray(sim["frequencies"])
frequency_scaling = np.asarray(sim["frequency_scaling"])
for frequency, scaling in zip(frequencies, frequency_scaling):
    print(f"{frequency:5.0f} GHz: dust amplitude a = {scaling:.4f} of its 353 GHz value")
print("data E-mode array shape (n_freq, nside, nside//2+1):", sim["data"].polar_matrix_1.shape)
```

The result is a dict. Its fields are `cmb_lensing` `FlatS2` objects in Fourier space, with E in `polar_matrix_1` and B in `polar_matrix_2`:

- `data` holds all three channels on a leading frequency axis.
- `cmb_field` and `dust_field` are the true $f$ and $g$ (single channel; the dust is at 353 GHz).
- `noise_covariance`, `scalar_covariance`, `tensor_covariance`, `field_covariance` and `dust_covariance` are `DiagonalEB` operators, and `mask` / `beam` are identities.

## What the model puts at each multipole

The covariances are diagonal in Fourier space, and each entry is the spectrum at that mode's multipole divided by the pixel area: the matrices hold $C_\ell / \theta_{\rm pix}^2$ with $\theta_{\rm pix}$ in radians (`pix_width`), the convention of `cmb_lensing`'s `covar_matrix_from_cls`. Reading one row of the grid and multiplying by `pix_width**2` therefore recovers each component's B-mode $C_\ell$, plotted here as $D_\ell = \ell(\ell+1)C_\ell / 2\pi$.


```python
INK = "#0b0b0b"
MUTED = "#898781"
GRID = "#e1e0d9"
SERIES = ["#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#e87ba4", "#008300"]

ell_grid, pix_width = gen_ell_grid(nside, theta_pix)
ells = np.asarray(ell_grid)[0, 1:]

def bb_dl(operator, channel = None):
    #D_ell of an operator's BB block along the first row of the rfft2 grid, in uK^2
    matrix = np.asarray(operator.matrix_BB)
    matrix = matrix if channel is None else matrix[channel]
    return ells * (ells + 1) * matrix[0, 1:] * pix_width**2 / (2 * np.pi)

fig, ax = plt.subplots(figsize = (8, 5))
ax.loglog(ells, bb_dl(sim["tensor_covariance"]) * r, color = SERIES[0], lw = 2,
          label = f"tensor, r = {r}")
ax.loglog(ells, bb_dl(sim["scalar_covariance"]), color = SERIES[1], lw = 2, label = "lensing")
for channel, (frequency, scaling) in enumerate(zip(frequencies, frequency_scaling)):
    if channel == len(frequencies)-1:
        continue
    ax.loglog(ells, bb_dl(sim["dust_covariance"]) * scaling**2, color = SERIES[2 + channel],
              lw = 2, ls = "--", label = f"dust, {frequency:.0f} GHz")
ax.loglog(ells, bb_dl(sim["noise_covariance"], channel = 0), color = SERIES[5], lw = 2,
          ls = ":", label = f"noise, {uk_arcmin:g} uK-arcmin")

#the lowest multipole the fine-pixel patch would have reached
_, ell_min_fine, _ = patch_multipoles(nside, theta_pix_fine)
_, ell_min_coarse, _ = patch_multipoles(nside, theta_pix)
ax.axvline(ell_min_fine, color = MUTED, lw = 1)
ax.text(ell_min_fine * 1.05, 0.97, f"min ell for {theta_pix_fine}' pixels", color = INK,
        fontsize = 9, va = "top", transform = ax.get_xaxis_transform())
ax.axvline(ell_min_coarse, color = MUTED, lw = 1)
ax.text(ell_min_coarse * 1.05, 0.97, f"min ell for {theta_pix}' pixels", color = INK,
        fontsize = 9, va = "top", transform = ax.get_xaxis_transform())

ax.set_xlabel(r"multipole $\ell$")
ax.set_ylabel(r"$D_\ell^{BB}$  [$\mu K^2$]")
ax.set_title("B-mode power of each component of the model", loc = "left")
ax.grid(color = GRID, lw = 0.6)
ax.set_axisbelow(True)
for side in ("top", "right"):
    ax.spines[side].set_visible(False)
ax.legend(frameon = False, fontsize = 9, ncols = 2, loc = "lower right")
plt.show()
```

Everything to the left of the grey line is only available because of the coarse pixels. The tensor curve peaks at the recombination bump near $\ell \approx 80$ and drops below the lensing curve beyond $\ell \approx 70$. Dust exceeds the tensor signal at the lowest multipoles even at 95 GHz, and by about two orders of magnitude at 220 GHz, which is why several channels are needed to separate it.

## The simulated maps

A Fourier-space matrix becomes a map through `irfft2`. First the two sky components on their own.


```python
DIVERGING = "RdBu_r"
width_deg, _, _ = patch_multipoles(nside, theta_pix)
extent = (0, width_deg, 0, width_deg)

def to_map(matrix):
    #real-space map (uK) of an rfft2 matrix; works with a leading frequency axis
    return np.fft.irfft2(np.asarray(matrix))

def show_map(ax, values, limit, title):
    #a map on a colour scale symmetric about zero
    image = ax.imshow(values, origin = "lower", extent = extent, cmap = DIVERGING,
                      vmin = -limit, vmax = limit)
    ax.set_title(title, loc = "left", fontsize = 10)
    return image

components = (("CMB", sim["cmb_field"]), ("dust at 353 GHz", sim["dust_field"]))
fig, axes = plt.subplots(2, 2, figsize = (9, 8), sharex = True, sharey = True, constrained_layout = True)
for row, (name, field) in enumerate(components):
    for column, (mode, matrix) in enumerate((("E", field.polar_matrix_1),
                                             ("B", field.polar_matrix_2))):
        values = to_map(matrix)
        image = show_map(axes[row, column], values, np.abs(values).max(), f"{name}, {mode}")
        label = ""
        if column == 1:
            label = r"$\mu K$"
        fig.colorbar(image, ax = axes[row, column], shrink = 0.85, label = label)
for ax in axes[-1]:
    ax.set_xlabel("degrees")
for ax in axes[:, 0]:
    ax.set_ylabel("degrees")
plt.show()
```

Each panel has its own colour scale, because the components differ by orders of magnitude: the CMB E-mode is several $\mu K$, the CMB B-mode a few tenths, and the dust at 353 GHz is much brighter than either. The dust maps are dominated by the largest scales, as its red spectrum implies.

Now the data: the three channels a telescope would actually record.


```python
data_maps = {"E": to_map(sim["data"].polar_matrix_1), "B": to_map(sim["data"].polar_matrix_2)}

fig, axes = plt.subplots(2, 3, figsize = (12, 7.5), sharex = True, sharey = True, constrained_layout = True)
for row, (mode, maps) in enumerate(data_maps.items()):
    #one colour scale per row so the channels can be compared directly
    limit = np.abs(maps).max()
    for channel, frequency in enumerate(frequencies):
        if channel > 2:
            continue
        image = show_map(axes[row, channel], maps[channel], limit,
                         f"data, {mode}, {frequency:.0f} GHz")
    fig.colorbar(image, ax = axes[row], shrink = 0.85, label = r"$\mu K$")
for ax in axes[-1]:
    ax.set_xlabel("degrees")
for ax in axes[:, 0]:
    ax.set_ylabel("degrees")
plt.show()
```

Within a row the colour scale is shared. The E-mode maps look almost identical across channels because the CMB E-mode dominates them and the CMB does not change with frequency. The B-mode maps change visibly: the large-scale dust pattern strengthens from 95 to 220 GHz, on top of pixel-scale noise. That frequency dependence is what lets the dust be separated from a primordial B-mode signal.

Change `master_seed` for a different realization, or `nside` and `theta_pix` to trade sky area against resolution.


## Non-Gaussian dust from PySM

Real Galactic dust is not a Gaussian random field: it is filamentary, its brightness varies strongly across the sky, and its Fourier modes are correlated. `simulate_data(..., dust_model = "pysm")` replaces the Gaussian draw of $g$ with dust from [PySM 3](https://pysm3.readthedocs.io/en/latest/models/), projected onto the flat patch. Only $g$ changes. The CMB, the noise and the frequency scaling $a_\nu$ are made exactly as above, and the four dust amplitude / knee arguments are not used. It needs the optional dependencies (`pip install -e ".[pysm]"`), and the first call downloads PySM's dust template.

The default model is `d9`: the Planck GNILC dust templates at 353 GHz with a single spectral index and temperature over the whole sky. `pysm_dust_field` turns it into a flat-sky field in five steps:

1. Evaluate the model at 353 GHz in $\mu K_{\rm CMB}$, giving Stokes I, Q and U on the full sphere (a healpix map in Galactic coordinates).
2. Mask the Galactic plane.
3. Split Q and U into E and B on the sphere, keep only the multipoles the flat grid can hold, and write E and B as two scalar maps.
4. Read those two maps at the pixels of the flat patch.
5. Subtract the patch mean and taper the edges to zero, so the cut-out is periodic like every other field in the model.

The cells below show each step.

### One sky, one realization

**`d9` has no seed.** It is a fixed map of the sky as Planck measured it, so every call returns the same dust and there is exactly one realization to work with. `master_seed` still changes the CMB and the noise, but not the dust. The only ways to get different PySM dust are to move the patch centre to another part of the sky or to switch model (PySM's `d11` redraws the small scales from a seed, but its large scales, which are the ones that matter for $r$, are still the one observed sky). A statement about a bias on $r$ from PySM dust can therefore average over CMB and noise realizations, but not over dust realizations.

### The full-sky map, the mask and the patch

The quantity shown is the polarized intensity $P = \sqrt{Q^2 + U^2}$ at 353 GHz, on a logarithmic colour scale, with the outline of the flat patch drawn on top. The patch is centred on the BICEP/Keck field (RA 0h, Dec $-57.5^\circ$).

```python
import healpy
import pysm3
import pysm3.units as units
from cmb_scattering.simulate_dust import (galactic_plane_mask, pysm_dust_eb_maps,
                                          flat_patch_sky_coordinates, edge_taper,
                                          DEFAULT_PYSM_PRESET, PYSM_DUST)

#the healpix resolution pysm_dust_field picks for 20' flat pixels (6.9' healpix pixels)
HEALPIX_NSIDE = 512
SEQUENTIAL = "magma"
#colour range of the polarized intensity maps, in uK_CMB
POLARIZED_INTENSITY_RANGE = (1.0, 1e4)
OUTLINE = "#1baf7a"

sky = pysm3.Sky(nside = HEALPIX_NSIDE, preset_strings = [DEFAULT_PYSM_PRESET],
                output_unit = units.uK_CMB)
stokes_maps = sky.get_emission(DUST_REFERENCE_FREQUENCY_GHZ * units.GHz).value
galactic_mask = galactic_plane_mask(HEALPIX_NSIDE)

def polarized_intensity(stokes):
    #sqrt(Q^2 + U^2) of an (I, Q, U) healpix map
    return np.sqrt(stokes[1]**2 + stokes[2]**2)

#Galactic coordinates of the pixels along the four edges of the flat patch
longitude, latitude = flat_patch_sky_coordinates(nside, theta_pix)
outline_longitude = np.concatenate([longitude[0], longitude[:, -1], longitude[-1, ::-1], longitude[::-1, 0]])
outline_latitude = np.concatenate([latitude[0], latitude[:, -1], latitude[-1, ::-1], latitude[::-1, 0]])

fig = plt.figure(figsize = (12, 4.2))
sky_maps = (("PySM d9 dust at 353 GHz, before the mask", polarized_intensity(stokes_maps)),
            ("after the Galactic plane mask", polarized_intensity(stokes_maps * galactic_mask)))
for panel, (title, sky_map) in enumerate(sky_maps):
    #the masked plane is exactly zero, which a log scale cannot show; it is drawn in grey
    healpy.mollview(np.where(sky_map > 0, sky_map, healpy.UNSEEN), fig = fig, sub = (1, 2, panel + 1),
                    norm = "log", min = POLARIZED_INTENSITY_RANGE[0], max = POLARIZED_INTENSITY_RANGE[1],
                    cmap = SEQUENTIAL, title = title, unit = r"polarized intensity [$\mu K_{\rm CMB}$]")
    healpy.projplot(outline_longitude, outline_latitude, lonlat = True, color = OUTLINE, lw = 2)
    healpy.graticule(dpar = 30, dmer = 60, color = MUTED, lw = 0.4)
plt.show()

in_patch = healpy.ang2pix(HEALPIX_NSIDE, longitude.ravel(), latitude.ravel(), lonlat = True)
print(f"patch centre: Galactic latitude {latitude[nside // 2, nside // 2]:.1f} deg; "
      f"its pixels span latitudes {latitude.min():.1f} to {latitude.max():.1f} deg")
print(f"polarized intensity: maximum on the sky {polarized_intensity(stokes_maps).max():.0f} uK, "
      f"median inside the patch {np.median(polarized_intensity(stokes_maps)[in_patch]):.0f} uK")
print(f"smallest mask value inside the patch: {galactic_mask[in_patch].min():.3f}")
```

The Mollweide projection is in Galactic coordinates: the Galactic plane is the horizontal band through the middle, and the green outline is the flat patch, in the southern Galactic hemisphere.

The mask is zero within $15^\circ$ of the plane and rises as a cosine ramp to one at $30^\circ$. PySM has no switch for this; the map it returns is the whole sky, plane included, and any masking is left to the user. The mask is needed even though the patch never touches the plane, because of step 3: E and B are *non-local* functions of Q and U. The scalar E map at one point depends on the polarization everywhere, and the plane is about a thousand times brighter than the patch. Unmasked, the plane's E-mode pattern spills across the whole sphere and dominated the patch.

### The flat patch, before and after the mean subtraction and taper

After the E / B split, the two scalar maps are read at the flat pixel positions (bilinear interpolation between healpix pixels). The left column below is that raw cut-out. It is not periodic: its left edge does not match its right edge, and an FFT would treat the jump as a sharp feature and spread spurious power over every scale. So the patch is multiplied by a Tukey window that is one in the middle and falls to zero over the outer 10 % on each side (20 % of the width in total). The mean is subtracted first (weighted by the window), because the window would otherwise turn a constant offset into a patch-sized plateau, which is spurious large-scale power. The right column is the dust field `simulate_data` returns.

```python
sim_pysm = simulate_data(nside, theta_pix, master_seed, r, frequency_exponent,
                         uk_arcmin = uk_arcmin, dust_model = PYSM_DUST)

#steps 3 and 4 by hand: the scalar E and B healpix maps (the multipoles between the flat
#grid's fundamental and its Nyquist), read at the flat pixels
lmin = int(2 * np.pi / (nside * pix_width))
lmax = int(np.pi / pix_width)
healpix_eb_maps = pysm_dust_eb_maps(DEFAULT_PYSM_PRESET, HEALPIX_NSIDE, lmin, lmax)
projected = {mode: healpy.get_interp_val(healpix_map, longitude, latitude, lonlat = True)
             for mode, healpix_map in zip(("E", "B"), healpix_eb_maps)}
#step 5 is done inside simulate_data
tapered = {"E": to_map(sim_pysm["dust_field"].polar_matrix_1),
           "B": to_map(sim_pysm["dust_field"].polar_matrix_2)}

fig, axes = plt.subplots(2, 2, figsize = (9, 8), sharex = True, sharey = True, constrained_layout = True)
for row, mode in enumerate(("E", "B")):
    #one colour scale per row, so the two columns can be compared directly
    limit = np.abs(projected[mode]).max()
    show_map(axes[row, 0], projected[mode], limit, f"PySM dust, {mode}: projected patch")
    image = show_map(axes[row, 1], tapered[mode], limit,
                     f"PySM dust, {mode}: mean removed and tapered")
    fig.colorbar(image, ax = axes[row], shrink = 0.85, label = r"$\mu K$")
for ax in axes[-1]:
    ax.set_xlabel("degrees")
for ax in axes[:, 0]:
    ax.set_ylabel("degrees")
plt.show()

taper = edge_taper(nside)
for mode in ("E", "B"):
    patch_mean = np.sum(projected[mode] * taper) / np.sum(taper)
    print(f"{mode}: window-weighted mean removed {patch_mean:+.2f} uK; "
          f"rms {projected[mode].std():.1f} uK before, {tapered[mode].std():.1f} uK after")
```

Compared with the Gaussian dust maps earlier in this guide, the PySM dust is visibly not statistically uniform: it is several times brighter along the top edge of the patch, the side closest to the Galactic plane, and its structure there is coherent over ten degrees or more, where a Gaussian field with the same spectrum would look the same everywhere. The taper fades the outer 10 % of each side, which lowers the total dust power a little.

### Same dust on every call

Two simulations that differ only in `master_seed` have different CMB and noise but identical dust:

```python
other_seed = master_seed + 1
sim_other_seed = simulate_data(nside, theta_pix, other_seed, r, frequency_exponent,
                               uk_arcmin = uk_arcmin, dust_model = PYSM_DUST)

def largest_difference(field_a, field_b):
    #largest absolute difference between the B maps of two fields, in uK
    return np.abs(to_map(field_a.polar_matrix_2) - to_map(field_b.polar_matrix_2)).max()

print(f"seeds {master_seed} and {other_seed}, largest difference between the B maps:")
print(f"  dust: {largest_difference(sim_pysm['dust_field'], sim_other_seed['dust_field']):.2e} uK")
print(f"  CMB:  {largest_difference(sim_pysm['cmb_field'], sim_other_seed['cmb_field']):.2e} uK")
print("dust covariance of a PySM simulation:", sim_pysm["dust_covariance"])
```

There is no `dust_covariance` for PySM dust, and no true value of the four dust power-law parameters: only `r` and `frequency_exponent` have a truth. [`03_sampling_pysm_dust.ipynb`](03_sampling_pysm_dust.ipynb) fits a power law to this dust and runs the Gaussian sampler on it.

A few approximations to keep in mind: the patch is laid out as a longitude / latitude grid about its centre, which distorts a $43^\circ$ patch by up to about 7 % at its edges; the dust has no power above the grid's Nyquist multipole along an axis ($\ell = 540$), while the corners of the Fourier grid reach higher; and the bilinear lookup smooths the smallest scales slightly.
